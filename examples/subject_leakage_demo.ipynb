# Subject-level leakage demonstration
This notebook uses simulated repeated observations. It does not reproduce the study cohorts or numerical results.


```python
import numpy as np
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from leakcheck import audit

rng = np.random.default_rng(7)
n_subjects, repeats, features = 40, 4, 20
subjects = np.repeat(np.arange(n_subjects), repeats)
labels = np.repeat(np.arange(n_subjects) % 2, repeats)
identity = rng.normal(size=(n_subjects, features))
X = identity[subjects] + 0.25 * rng.normal(size=(len(subjects), features))
```

```python
result = audit(
    estimator_factory=lambda: make_pipeline(StandardScaler(), LogisticRegression(max_iter=2000)),
    X=X,
    y=labels,
    subjects=subjects,
    n_folds=4,
    seeds=range(5),
)
result
```
